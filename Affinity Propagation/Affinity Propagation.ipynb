```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
dataset
```

```python
X = dataset.iloc[:,3:5].values
#X = dataset.iloc[:,[3,4]].values
# : -- It means it will select all the rows
# [3,4] -- It means it will select 3rd and 4th column alone
```

```python
X
```

```python
from sklearn.cluster import AffinityPropagation

clustering = AffinityPropagation()
lable = clustering.fit_predict(X)
```

```python
# from sklearn.cluster import KMeans
# kmeans = KMeans(n_clusters = 5, init = 'k-means++', random_state = 42)
# y_kmeans = kmeans.fit_predict(X)
```

```python
lable
```

```python
supervised=dataset
supervised['Cluster_group']=lable
```

```python
supervised
```

```python
supervised.to_csv("Affinity Propagation Cluster.csv",index=False)
```

```python
import seaborn as sns
facet = sns.lmplot(data=supervised, x=supervised.columns[3], y=supervised.columns[4], hue=supervised.columns[5],legend=True, legend_out=True)
```
