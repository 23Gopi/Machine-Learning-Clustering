```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
dataset
```

```python
X = dataset.iloc[:,3:5].values
#X = dataset.iloc[:,[3,4]].values
# : -- It means it will select all the rows
# [3,4] -- It means it will select 3rd and 4th column alone
```

```python
X
```

```python
from sklearn.cluster import MeanShift
mean_shift = MeanShift(bandwidth=25)
y_kmeans = mean_shift.fit_predict(X)

# Bandwidth → controls the size of the search area 
# Small bandwidth → more clusters 
# Large bandwidth → fewer clusters  
```

```python
y_kmeans
```

```python
supervised=dataset
supervised['Cluster_group']=y_kmeans
```

```python
supervised
```

```python
supervised.to_csv("Mean Shift Cluster.csv",index=False)
```

```python
import seaborn as sns
facet = sns.lmplot(data=supervised, x=supervised.columns[3], y=supervised.columns[4], hue=supervised.columns[5],legend=True, legend_out=True)
```
